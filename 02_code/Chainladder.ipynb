```python
import pandas as pd

cl_results = pd.read_excel('../03_chainladder/ChainLadder.xlsx', sheet_name="Input_data")
```

```python
cl_results_display = cl_results[
    [
        "Occurrence Period",
        "Paid",
        "Ultimate",
        "% Developed",
        "CL Ultimate",
        "Reserve Bias %",
        "Reserve Bias"
    ]
]

cl_results_formatted = cl_results_display.copy()

amount_columns = [
    "Paid",
    "Ultimate",
    "CL Ultimate",
]

for column in amount_columns:
    cl_results_formatted[column] = (
        cl_results_formatted[column]
        .map(lambda value: f"{value:,.2f}")
    )

cl_results_formatted["% Developed"] = (
    cl_results_formatted["% Developed"]
    .map(lambda value: f"{value:.2%}")
)

cl_results_formatted["Reserve Bias %"] = (
    cl_results_formatted["Reserve Bias %"]
    .map(
        lambda value: (
            f"{value:.2%}"
            if pd.notna(value)
            else "N/A"
        )
    )
)

cl_results_formatted["Reserve Bias"] = (
    cl_results_formatted["Reserve Bias"]
    .map(lambda value: f"{value:,.2f}")
    )

cl_results_formatted
```

```python
import matplotlib.pyplot as plt

plt.figure()

cl_results = cl_results[cl_results["Occurrence Period"] != "Total"]
# plt.plot(cl_results["Occurrence Period"], cl_results["Ultimate"])
# plt.plot(cl_results["Occurrence Period"], cl_results["CL Ultimate"])


fig, ax = plt.subplots()
ax.plot(cl_results["Occurrence Period"], cl_results["Ultimate"], linestyle='--', label='Actual')
ax.plot(cl_results["Occurrence Period"], cl_results["CL Ultimate"], linestyle='--', label='Predicted')
ax.set_yscale("log") 
ax.set_xlabel('Occurrence period', fontsize=15)
ax.set_ylabel('Total Ultimate claims', fontsize=15)
ax.set_title('by Occurrence Period')     
ax.legend()
# writer.add_figure('by Occur Period', fig)
```
